Load csv files

```python
australia = pd.read_csv('Australia_Wine_Stats.csv')
chile = pd.read_csv('Chile_Wine_Stats.csv')
france = pd.read_csv('France_Wine_Stats.csv')
italy = pd.read_csv('Italy_Wine_Stats.csv')
new_zealand = pd.read_csv('New Zealand_Wine_Stats.csv')
portugal = pd.read_csv('Portugal_Wine_Stats.csv')
spain = pd.read_csv('Spain_Wine_Stats.csv')
usa = pd.read_csv('USA_Wine_Stats.csv')
```

Combining the 8 csv files into a single data frame

```python
wine_df = pd.concat([australia, chile, france, italy, new_zealand, portugal, spain, usa], ignore_index=True)
```
